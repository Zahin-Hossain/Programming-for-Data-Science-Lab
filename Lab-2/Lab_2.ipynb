```python
#Problem 1
student_name = input("Enter student's name:")
marks = [
    float(input("Enter marks for course 1:")),
    float(input("Enter marks for course 2:")),
    float(input("Enter marks for course 3:")),
    float(input("Enter marks for course 4:")),
    float(input("Enter marks for course 5:"))
]

total=sum(marks)

print("Total marks:",total)
print("Average marks:",total/5)
print("Highest marks:",max(marks))
print("Lowest marks:",min(marks))

passed=0
for i in marks:
    if i>=50:
        passed+=1

print("Number of courses passed:",passed)
```

```python
#Problem 3
text=input("Enter a word/short sentence:")

characters=0
vowels=0
spaces=0

for char in text:
    characters+=1

    if char in "aeiouAEIOU":
        vowels+=1

    if char==" ":
        spaces+=1

print("Number of characters:",characters)
print("Number of vowels:",vowels)
print("Number of spaces:",spaces)
```

```python
#Problem 4
def calculate_grade(marks):
    average=sum(marks)/len(marks)

    if average>=80:
        grade="A"

    elif average>=70:
        grade="B"

    elif average>=60:
        grade="C"

    elif average>=50:
        grade="D"


    else:
        grade="F"

    return average,grade

marks=[]

for i in range(5):
    mark=float(input("Enter mark for course "+str(i+1)+":"))
    marks.append(mark)

average,grade=calculate_grade(marks)

print("Average:",average)
print("Grade:",grade)
```

```python
#Problem 5
products = [
    {"name": "Apple", "price": 100, "quantity": 5},
    {"name": "Banana", "price": 60, "quantity": 10},
    {"name": "Mango", "price": 150, "quantity": 8},
    {"name": "Orange", "price": 80, "quantity": 5}
]

total=0

for product in products:
    total=total+product["price"]*product["quantity"]

print("Total:",total)

def apply_discount(total):
    if total>=2000:
        total=total-(total*0.10)

    return total

final_total=apply_discount(total)

print("Final total:",final_total)
```

```python
#Problem 7
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total_marks=0
passed=0
highest=0
top_student=""

for student in students:
    print(student["name"],"-",student["marks"])

    total_marks=total_marks+student["marks"]

    if student["marks"]>=50:
        passed+=1

    if student["marks"]>highest:
        highest=student["marks"]
        top_student=student["name"]

average=total_marks/len(students)

print("Students passed:",passed)
print("Class average:",average)
print("Highest mark:",highest)
print("Top student:",top_student)
```

```python
#Problem 8
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total_value=0
highest_total=0
highest_product=""

for product in products:
    line_total=product["price"]*product["quantity"]

    print(product["name"],"-",line_total)

    total_value=total_value+line_total

    if line_total>highest_total:
        highest_total=line_total
        highest_product=product["name"]

print("Total inventory value:",total_value)
print("Highest line total product:",highest_product)
print("Highest line total:",highest_total)
```
